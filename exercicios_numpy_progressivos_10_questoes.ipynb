## Exercício 6 — Índice do maior valor

Considere as vendas mensais:

```python
vendas = np.array([3200, 4100, 3900, 5200, 4800, 6100])
```

Descubra:

1. qual foi a maior venda;
2. em qual índice ela aparece;
3. qual foi a menor venda;
4. em qual índice ela aparece.

Use `np.argmax()` e `np.argmin()`.

````python
vendas = np.array([3200, 4100, 3900, 5200, 4800, 6100])

maior = np.argmax(vendas)
menor = np.argmin(vendas)

print("Maior venda:", vendas[maior])
print("Índice da maior venda:", maior)

print("Menor venda:", vendas[menor])
print("Índice da menor venda:", menor)
````

# Dicas rápidas

Alguns comandos que podem ser úteis:

```python
np.sum(array)
np.mean(array)
np.max(array)
np.min(array)
np.argmax(array)
np.argmin(array)
np.sum(array > valor)
array[array > valor]
np.sum(matriz, axis=0)
np.sum(matriz, axis=1)
np.mean(matriz, axis=0)
np.mean(matriz, axis=1)
np.unravel_index(indice, matriz.shape)
```

Lembre-se:

- `axis=0` produz um resultado por coluna;
- `axis=1` produz um resultado por linha;
- máscaras booleanas podem ser usadas diretamente para filtrar arrays;
- operações matemáticas podem ser aplicadas diretamente ao array inteiro.
